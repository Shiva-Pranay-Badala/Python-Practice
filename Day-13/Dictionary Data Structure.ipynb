```python
int_dict = {1: 'mamidipandu', 2 : 'Aratipandu', 3: 'Batai'}  # keys are defined as int
print(int_dict)

float_dict = {1.2 : 'mamidipandu', 2.3 : 'Aratipandu', 3.4 : 'Batai'} # keys are defined as float
print(float_dict)

string_dict = {'mango' : 'mamidipandu', 'banana' : 'Aratipandu', 'orange' : 'Batai'} # keys are defined as string
print(string_dict)

hetro_dict = {1 : 'mamidipandu', 'banana' : 'Aratipandu', 3.4 : 'Batai'} # keys are defined as hetro
print(hetro_dict)

tuple_dict = {(10,20) : 'mamidipandu', 'banana' : 'Aratipandu', 3.4 : 'Batai'} # keys are defined as tuple
print(tuple_dict)

# tuple can be used as the key when u use multiple values as a key we  use tuple
#list & set cannot be used as the key
#keys must be immutable due to hash table

# values can be mutable and immutable

dict_comp = {x : x**2 for x in range(1,6)}
print(dict_comp)
```

```python
#coverting to dictionary

list_of_tuple = [('Gopichand',6),('chandar',90),('Dinesh',100)]

dict1 = dict(list_of_tuple)
print(dict1)

list_of_list = [['Gopichand',6],['chandar',90],['Dinesh',100]]
dict2 = dict(list_of_list)
print(dict2)

tuple_of_tuple = (('Gopichand',6),('chandar',90),('Dinesh',100))
dict3 = dict(tuple_of_tuple)
print(dict3)

tuple_of_list = (['Gopichand',6],['chandar',90],['Dinesh',100])
dict4 = dict(tuple_of_list)
print(dict4)

set_of_tuple = {('Gopichand',6),('chandar',90),('Dinesh',100)}
dict5 = dict(set_of_tuple)
print(dict5)
```
