# Inheritance

```python
# Parent class = Base class
# Child class = Derived class
```

# Types of Inheritance

# 1. Single inheritance : One parent on child


```python
class hyd_food:
    taste = "very very very Spicy"

    def Biriyani(self):
        print("Very very tasty Biriyani")

# child class
class PistaHouse(hyd_food):
    pass

obj_child = PistaHouse() # it is the object of child class

# child class object can access the properties & function of parent class
print(obj_child.taste)
obj_child.Biriyani()
```

```python
# example 2 super() function
# super() : used aceess the parent properties and function from inside of the class

class hyd_food:
    taste = "very very very Spicy"

    def Biryani(self):
        print("Very very tasty Biryani")

class PistaHouse(hyd_food):
    def display(self):
        super().Biryani()
        print(super().taste)

obj_child = PistaHouse()
obj_child.display()

```

```python
class hyd_food:
    taste = "very very spicy"
    def __init__(self):  # constructor is used in child class
        print("i am parent class")

        # child class
class Pistahouse(hyd_food):
    def __init__(self): # constructor is used in child class
        print("i am child class")

obj_child = Pistahouse()  # it is the object of child class
obj_child = hyd_food()

# if child class constructor function is not defined
# them child class object will call automatically parent class function
```

# 2. Multiple inheritance : many parents one child


```python
# Example-1 : design the structure
class A:
    def display(self):
        print("I am A class")
class B:
    def show(self):
        print("I am B class")

class C(A,B):
    pass

obj_C = C()
obj_C.display()
obj_C.show()
```

```python
# Example-2 : Sequence in which  parent are assingn matters
```

```python
# Example 3 :constructor function

class A:
    def __init__(self):
        print("I am A class Constructor")

class B:
    def __init__(self):
        print("I am B class Constructor")
        
class C(B,A):
    def __init__(self):
        print("I am C class Constructor")
    

obj_C = C()
    
```

```python
# Example-2 : super()

class A:
    def display(self):
        print("I am A class Constructor")

class B:
    def show(self):
        print("I am B class Constructor")
        
class C(A,B):
    def hello(self):
        super().display()
        super().show()
    

obj_C = C()
obj_C.hello()
```

```python
class A:
    def display(self):
        print("I am A class Constructor")

class B:
    def show(self):
        print("I am B class Constructor")
        
class C(A,B):
    def hello(self):
        super().display()
    

obj_C = C()
obj_C.hello()

# MRO : method resolution Order
print([i.__name__ for i in C.__mro__])
```

# Multilevel Inheritance


```python
# son/daughter -> father -> grandfather -> grandGF
```

```python
# Example-1: Structure of the multi inheritance
class grandGF:
    def Gold(self):
        print("1000 KG")

class grandfather(grandGF):
    def Land(self):
        print("500 Acr")

class father(grandfather):
    def Car(self):
        print("Lamborgini")

class son(father):
    def Bike(self):
        print("RE bike")
        
obj_son = son()
obj_son.Bike()
obj_son.Car()
obj_son.Land()
obj_son.Gold()
```

```python
# Example-2: Constructor functions
class grandGF:
    def __init__(self):
        print("1000 KG")

class grandfather(grandGF):
    def __init__(self):
        print("500 Acr")

class father(grandfather):
    def __init__(self):
        print("Lamborgini")

class son(father):
    def __init__(self):
        print("RE bike")

obj_son = son()     
```

```python
# Example-2: Super()
class grandGF:
    def Land(self):
        print("1000 KG")

class grandfather(grandGF):
    def Land(self):
        print("500 Acr")

class father(grandfather):
    def Car(self):
        print("Lamborgini")

class son(father):
    def Bike(self):
        print("RE bike")
        super().Car()
        super().Land()
        super().Gold()
        
        
obj_son = son()
obj_son.Land()
```

```python
# Example 4: Method overriding
# Def: if child class is having same name function as the parent class then
# child class function will OVERRIDE the function of parent class

class grandGF:
    def Land(self):
        print("1000 KG")

class grandfather(grandGF):
    '''
    def Land(self):
        print("500 Acr")
    '''

class father(grandfather):
    '''
    def Car(self):
        print("Lamborgini")
    '''

class son(father):
    '''
    def Bike(self):
        print("RE bike")
    '''

obi_son = son()
obj_son.Car()
        
```

# Hierarical Inheritance


```python
# Many child One Parent

# Example-1 : Structure of Hierarical inheritance

class Superhero:
    def intro(self):
        print("Superhero powers")

class Sinchain(Superhero):
    def comedy(self):
        print("Himavari")

class Doreman(Superhero):
    def pocket(self):
        print("Nobita friend")

# Example 1: Using Doreman
obj_doreman = Doreman()
obj_doreman.intro()
obj_doreman.pocket()

# Example 2: Using Sinchain
obj_sinchain = Sinchain()
obj_sinchain.intro()
obj_sinchain.comedy()
```

```python
# Example-3: Super() function
class Superhero:
    def intro(self):
        print("Superhero powers")

class Sinchain(Superhero):
    def comedy(self):
        print("Himavari")

class Doreman(Superhero):
    def pocket(self):
        print("Nobita friend")
        super().intro()

obj_doreman = Doreman()
obj_doreman.pocket()

# Method resolution order
print([i.__name__ for i in sinchain.__mro__])
```

# Hybrid inheritance


```python
# example-1: structure of hybrid inheritance
class A:
    A_var = "A Class"

class B(A):
    B_var = "B Class"

class C(B):
    C_var = "C Class" 

class D(B):
    D_var = "D Class"

class E(C,D):
    E_var = "E Class"

obj_E = E()
print(obj_E.A_var)
    
```

```python
# Example-3:

class A:
    A_var = "A Class"

    def A_output(self):
        print("This is A output")

class B(A):
    B_var = "B Class"

    def B_output(self):
        print("This is B output")


class C(B):
    C_var = "C Class" 

    def C_output(self):
        print("This is C output")


class D(B):
    D_var = "D Class"

    def D_output(self):
        print("This is D output")


class E(C,D):
    E_var = "E Class"

    def E_output(self):
        print("This is E output")
        super().D_output()
        super().C_output()
        super().B_output()

        
obj_E = E()
obj_E.E_output()

print([i.__name__ for i in E.__mro__])
    
```
