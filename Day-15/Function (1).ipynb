# function

```python
# Def: A function is a block of reusable code that perform a specific task.

def fun():   # definig the function
    print("Welcome to functions")

    print("Hello world")

fun()  # calling the function until and unless we dont calll the function it will not extecute
fun()  # u call the function as many times as u want
```

```python
# ex-2 : passing one parameter in the function

def fun(val):   # definig the function and passing value to a function
    print("Welcome to functions", val)

    print("Hello world")

fun(100)  # calling the function
```

```python
# ex-3 : passing parameter in the function

def fun(val1, val2):   # definig the function and passing value to a function
    print("Welcome to functions", val1)

    print("Hello world", val2)

fun(100,200)  # calling the function
```

```python
# ex-4 : nested function : function inside the function (we can only do in python not in other languages)

def outer():
    print("Iam outer function")
    def inner():
        print("Iam inner function")
    inner()
outer()
```

```python
def bulding():
    print("we are here")
    def room1():
        print("pavan: iam here")
    room1()  # calling from inside the bulding function
bulding()
```

```python
# ex-5 : Return function

def fun(a,b):
    c = a + b
    return c    # return single value
    
val = fun(10,20)
print(val)
```

```python
def fun(a,b):
    c = a + b
    d = a - b
    e = a * b
    return c,d,e   # returing multiple values
    
val = fun(20,10)
print(val)
```
