# OOP: Object Oriented Programming


```python
# EX-1: Structure of Class

#class is a keyword, foood is the class name
# class variable: the value of class variable will remains SAME for all the objects
# class variable value cannot be change by object but it can only change by using class name
class food:    
    food_name = "Biryani"   # class variable we defined

obj = food()    # creating the object (obj) of food() class
obj.food_name  # accessing the class variable
```

```python
# we can create multiple objects
class food:    
    food_name = "Biryani"   # class variable we defined

obj1 = food()    # creating the object (obj) of food() class
print(obj1.food_name)  # accessing the class variable

obj2 = food()
print(obj2.food_name) # class object (obj2) can access the class data

# we cannot access class variable diectly we have to access by using object
```

```python
# access the class variable with class name
print(food.food_name)
```

```python
# class variable value cannot be change by object
# class variable CAN be only change by class name
class food:    
    food_name = "Biryani"   # class variable we defined

obj1 = food()    # creating the object (obj) of food() class
print(obj1.food_name)  # accessing the class variable


food.food_name = "Mutton" #class variable CAN be only change by class name

obj2 = food()
print(obj2.food_name) # class object (obj2) can access the class data
```
