# Encapusulation


```python
# Def : Encapsulation refers to the binding of data (attributes) and functions into one unit(class).

# Encapsulation restrict the direct access of the data.

# We can access the variables/data/attributes through object outside the class.
# We can access the variables/data/attributes through functions defined inside the class.

class Bike:
    No_tyre = 2    # class variable / Attribute / data

    def __init__(self, name, price):
        self.bike_name = name     # instance variable
        self.price = price

    def display(self):
        print(self.No_tyre,self.bike_name,self.price)

obj1 = Bike("RE", 1.5)
obj1.display()

print(obj1.No_tyre)
print(obj1.bike_name)
print(obj1.price)
```

# Access specifiers / type of attributes in encapsulation


```python
class marvels:
    def __init__(self, name, power, weapon):
        self.Name = name   # public instance variable
        self._Power = power # Protected instance variable
        self.__Ayudham = weapon # Private instance variable

    def show(self):
        print(self.Name,self._Power,self.__Ayudham)

obj1 = marvels("Thor", 1000, "Hammer")
obj1.show()

print(obj1.Name)
print(obj1._Power)

# Procted variables can be access outside the class, but you should not access.
# Python public, procted, private concept is used just for sign/warning
print(obj1._marvels__Ayudham) # obj._classname__private variable
```
