# Default Parameters

```python
# assign a default value to a "var" variable
def fun(var = "Nothing"):
    print(var)

fun()
fun(100)  # here we are passing value
```

```python
def fun2(x = 10, y = 20):
    print(x + y)

fun2()
fun2(100)
fun2(100, 200)
```

# passing arbitary number of arguments using *args


```python
def fun(*args):
    for i in args:
        print(i)

fun()
fun("chinki")
fun("chinki","Minki") # u can ass any numbers of arguments to the function by using *args
```

```python
def fun(*var):
    for i in var:
        print(i)

fun()
fun("chinki")
fun("chinki","Minki")
```
