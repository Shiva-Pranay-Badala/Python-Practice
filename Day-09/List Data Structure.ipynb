# creating different type of list

```python
empty_list = []       # by using []
empty_list2 = list()  # by using list function

print(empty_list, type(empty_list))
print(empty_list2, type(empty_list2))
```

```python
Het_list = [10, 20.5, "Hello", True]
print(Het_list)
```

```python
dup_list = [10, 10, 10, 40, 50]
print(dup_list)
```

```python
# nested list
nested_list = [[1,2],[10,20],["Hello","World"],1000]
print(nested_list)
print(len(nested_list))
print(nested_list[1][0])
```

```python
# concatenation on list
concat_list = dup_list + Het_list + nested_list
print(concat_list)
```

# Accessing list items

```python
Hours_study = [1,3,4,3,5,2,1.5,6,20,6,24]

print(Hours_study[4:8:1])
Hours_study[4:8:-1]
```

# Updating list items

```python
list_politician = ["vijay", "CBN", "Pavan", "JaiLalita", "Balayya", "Anushka"]

#update single element
list_politician[1] = "Dance With"

print(list_politician)
```

```python
# update multiple elements: update more elements with less
list_politician[3:5] = ["Dance With"]
print(list_politician)
```
