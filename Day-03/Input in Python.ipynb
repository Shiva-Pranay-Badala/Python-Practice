# 3. Assigment operators (+=, -=, *=, /=,...)

```python
a = 10

a+=5 # a = a + 5
print(a)
a-=5 # a = a -5
print(a)
a*=5
print(a)
a/=5
print(a)
a%=5
print(a)

# &=, |=, ^=, <<=, >>=, :=(walrus operator) ...

a = 10
a &= 3
print(a)

a = 10
a |= 3
print(a)

a = 10
a ^= 3
print(a)

a = 10
a <<= 3
print(a)

a = 10
a >>= 3
print(a)

#Walrus Operator (:=)
#Walrus Operator will first assign the value to the variable then pass it to outer function
#on the go u assign the value and then print the value

print(a := 3)
```

# 4. Comparison Operators(==, >=, <=, !=, <, >)

```python
a = 10
b = 5

print(a == b)
print(a >= b)
print(a <= b)
print(a != b)
print(a < b)
print(a > b)
```

# 5. Logical operators(and, or, not)

```python
#and, or, not
a = True
b = False

print( a and b)
print( a or b)
print( not b)
```

# 6. Identity Operator(is, is not)

```python
a = [1,2,3]
b = [1,2,3]
c = a

print(a == b)

print(a is b) #Its False because their memory address will be different
print(a is c) #Its True because c is assgined as a
print(a is not b)
print(a is not c)
```

# 7. Membership Operators (in, not in)

```python
fruits = ('Mamidipandu', 'Aratipandu', 'Batai', 'panasapandu')

print('Batai' in fruits)
print('Batai' not in fruits)
```

# Input in Python

```python
#input function is used to take input from the user


val = input("Enter your phone number")

print(val)
print(type(val)) #By default input function recieve/return the data in string format
```

```python
val = int(input("Enter your phone number"))

print(val, type(val)) # string format can be coverted into other type by using typecasting method
```

```python
# recieving multiple inputs we use split function( .split() )

x1, x2, x3 = input("Enter 3 names").split() # by default give multiple inputs by giving space
print(x1, x2, x3)
```

```python
x1, x2, x3 = input("Enter 3 names").split(",") #if u want to use special char mention that in bracket()
print(x1)
```

```python
#map()

x1, x2, x3 = map(int, input("Enter 3 num").split()) #for typecasting we use map
print(x1, x2, x3)
```

```python
x1, x2, x3 = map(float, input("Enter 3 names").split()) #for typecasting we use map
print(x1, x2, x3)
```
