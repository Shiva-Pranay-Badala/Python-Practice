# Recurise function

```python
# ex-1
def foo(i):
    if i < 1:
        return
    print(i)
    foo(i-1)    # recurise call
foo(6)
```

```python
# ex-2
def foo(i):
    if i < 1:  
        return
    foo(i-1)
    print(i)
foo(6)
```

```python
# ex-2
def foo(i):
    if i < 1:  # termination condition r base condition also must be there in recusion fun
        return
    print(i)
    foo(i-1)     # when ur calling recuserive fun u have to increment r decrement
    print(i)
foo(6)
```
