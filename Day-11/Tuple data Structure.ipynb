# creating different types of tuple


```python
empty_tuple = ()
empty_tuple2 = tuple()
print(empty_tuple,type(empty_tuple))
print(empty_tuple2,type(empty_tuple2))
```

```python
het_tuple = (10, 10.5, "hello",True)
print(het_tuple)
```

```python
dup_tuple = (10, 20, 30, 10, 10, 10,20)
print(dup_tuple)
```

```python
nested_tuple = (dup_tuple, het_tuple,empty_tuple)
print(nested_tuple)
```

```python
nested_tuple = ((10,20,30),[40,50],{80,90,60},"Hello")
print(nested_tuple )
```

```python
# convert list and set to tuple using tuple()
l1 = [10,20,30,40]
s1 = {30,40,50,60}
l1_t1 = tuple(l1)
s1_t1 = tuple(s1)
print(l1_t1)
print(s1_t1)
```

```python
values = range(1,11)
t1 = tuple(values)
print(t1)
print(tuple(range(1,5)))
```

```python
t1 = (1,2,3)
t2 = (4,5,6)
t3 = (7,8,9)
print(t1 + t2 + t3) # tuple concatination
```

```python
t5 = tuple(i for i in range(10))
print(t5)
```

# accessing the tuple items


```python
veg_tuple = ("tomato","Mirapakaya","alugada","vankaya","bendakaya","sorakaya")
veg_tuple[::2]
veg_tuple[-1:-10:-1] #reverse
veg_tuple[-1:0:-1]
# we access tuple by positive and negative indexing
```

# insert, delete, update operations indirectly


```python
t1 = (10,10.2,"Hello",[10,20,30])
t1[3][2]
```

```python
t1 = (10,10.2,"Hello",[10,20,30])
```

```python
t1[3][2] =700
```

```python
t1      # we can't update complete item in tuple but we can update element in the mutable item if it in tuple
```

```python
# tuple items cannot be updated directly
# complete item in tuple cannot be updated directly
# but if item is mutuable then elements inside this mutable item can be updated
```

```python
# if u to perform insert, delete, update operation indirectly first convert tuple to list then perform the operstions and then covert to tuple
```

```python
#insert list: append(),insert(),extend()
t1 = (10,20,30)
l1 = list(t1)
print(l1)
l1.append(500)
print(l1)
t2 = tuple(l1)
print(t2)
```

```python
#remove list: remove(),pop(),clear(),del
t1 = (10,20,30)
l1 = list(t1)
print(l1)
l1.remove(30)
print(l1)
t2 = tuple(l1)
print(t2)
```

```python
#update list: 
t1 = (10,20,30)
l1 = list(t1)
print(l1)
l1[1] = 60
print(l1)
t2 = tuple(l1)
print(t2)
```

# popular inbuilt function for tuple

```python
salary = (6, 70, 100, 600, 700, 90, 10, 80, 120)

print(tuple(reversed(salary)))
print(len(salary))
print(min(salary))
print(max(salary))
print(sum(salary))

sorted_list=tuple(sorted(salary)) # we can't apply sort function for tuple but we can sort by using sorted() -> it takes extra place to sort
sorted_list1=tuple(sorted(salary, reverse = True)) # we get list so we use tuple nd convert it to tuple
print(sorted_list)
print(sorted_list1)

# sort() is inplace function -> sorting is done in same data sturcture it doesn't take extra space

# sorted() function return the list
# sort() function don't return anything
```
