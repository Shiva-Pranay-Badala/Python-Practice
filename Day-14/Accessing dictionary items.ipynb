# Accessing dictionary items

```python
drink_items = {
                "Thumpsup" : 50,
                "Sugarcane" : 20,
                 "coconutwater" : 60,
                 "water" : 20,
                  "rosemilk" : 30
}

val = drink_items["water"]  # if value is not their it will show error
print(val)

val2 = drink_items.get("rosemilk")
print(val2)

val2 = drink_items.get("coke","key not avaliable") # if key is not their it will not show error
print(val2)
```

```python
#to access the keys
for i in drink_items:
    print(i)
```

```python
#to access the keys and values
for i in drink_items:
    print(i, drink_items[i])
```

```python
# to access all the keys we use key() functions
print(drink_items.keys())
```

```python
# to access all the values we use values() functions
print(drink_items.values())
```

```python
# to access all the items we use items() functions
print(drink_items.items())
```

```python
# nested dictionary
nested_dict = {
                100 : {"Hello":10, "World" : 20},
                200 : {"A" : 50 , "B" : 60}
}

nested_dict[100]
nested_dict[100]["World"]
```

# insert & update opeartions on dictionary


```python
x_dict = {"name" : "Chinki", "age" : 80}

x_dict["age"] = 180  # if key is avaliable it will update
print(x_dict)

x_dict["city"] = "Kurnool"  # if key is not avaliable it will insert
print(x_dict)
```

```python
# insert & update using update() function
# updat() can insert & update multiple items

x_dict.update({'age' : 18, 'occupation' : 'IAS', "name" : "minki", "Salary" : 1.5})
print(x_dict)
```

# delete the items from dictionary

```python
# pop(), popitem(), clear(), del()

x_dict
```

```python
# pop() removes the items whhose key is specified
x_dict.pop("Salary")
print(x_dict)
```

```python
# popitem() removes last  element
x_dict.popitem()
print(x_dict)
```

```python
# clear() clears all the elemments
x_dict.clear()
print(x_dict)
```

```python
# deletes whole dictionary
del x_dict
```
