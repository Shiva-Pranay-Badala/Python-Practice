# Decorators

```python
def decorator_fun(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun          # modified function

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

decorated_function = decorator_fun(cake)
decorated_function()
```

```python
def decorator_fun(func):   # func => cake()
    def wrapper_fun():# wrapper function add the extra functionality
        print("Buy ingrdents")   # we can add extra functionality before nd after the original fun
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun          # modified function

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

decorated_function = decorator_fun(cake)
decorated_function()
```

```python
# we repersent decorator with the help of @

# ex-2: use of @ to repersent the decorator

def decorator_fun(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        print("Buy ingrdents")
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun 

@decorator_fun        # if we use '@' with decorator function then it gets decorated
def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

cake()
```

```python
# multiple decorator for same function

def decorator_fun1(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        print("Buy ingrdents")
        func()
    return wrapper_fun

def decorator_fun2(func):   # func => cake()
    def wrapper_fun():   # wrapper function add the extra functionality
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun

@decorator_fun1
@decorator_fun2

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

cake()
```
